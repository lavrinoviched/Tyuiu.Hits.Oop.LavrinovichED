```polyglot-notebook
using System;
using System.Collections.Generic;

public abstract class Person
{
    public string Name { get; set; }
    public string Position { get; set; }
    
    public Person(string name, string position)
    {
        Name = name;
        Position = position;
    }
    
    public abstract void DisplayBasicInfo();
    
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}");
        Console.WriteLine($"Должность: {Position}");
    }
}

public class Employee : Person
{
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
        : base(name, position)  // ← вызов конструктора абстрактного класса
    {
        Salary = salary;
    }
    
    public override void DisplayBasicInfo()
    {
        Console.WriteLine($"{Name} - {Position}");
    }

    public override void PrintInfo()
    {
        base.PrintInfo();  
        Console.WriteLine($"Зарплата: {Salary:N2} руб.");
        Console.WriteLine(); 
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }
    
    public override void DisplayBasicInfo()
    {
        Console.WriteLine($"{Name} - {Position} (Отдел: {Department})");
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
        Console.WriteLine();
    }
}

public class ProjectManager : Manager
{
    public string ProjectName { get; set; }

    public ProjectManager(string name, string position, double salary, string department, string projectName)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }
    
    public override void DisplayBasicInfo()
    {
        Console.WriteLine($"{Name} - {Position} (Проект: {ProjectName})");
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
        Console.WriteLine();
    }
}

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class AdvancedProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    private string[] _teamMembers;

    public AdvancedProjectManager(string name, string position, double salary, 
                                string department, string projectName, string[] teamMembers)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        _teamMembers = teamMembers ?? new string[0]; 
    }
    
    public override void DisplayBasicInfo()
    {
        Console.WriteLine($"{Name} - {Position} (Проект: {ProjectName}, Команда: {_teamMembers.Length} чел.)");
    }

    public override void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary:N2} руб.");
        Console.WriteLine($"Отдел: {Department}");
        Console.WriteLine($"Проект: {ProjectName}");
        Console.WriteLine($"Размер команды: {_teamMembers.Length} человек");
        Console.WriteLine();
    }

    public string GenerateReport()
    {
        return $"Отчет по проекту '{ProjectName}'. " +
               $"Менеджер: {Name}. " +
               $"Команда: {_teamMembers.Length} человек. " +
               $"Статус: в работе.";
    }

    public string[] GetTeamMembers()
    {
        return _teamMembers;
    }

    public void DisplayTeam()
    {
        Console.WriteLine($"Команда проекта '{ProjectName}':");
        for (int i = 0; i < _teamMembers.Length; i++)
        {
            Console.WriteLine($" {i + 1}. {_teamMembers[i]}");
        }
    }
}

        Console.WriteLine("Система управления сотрудниками предприятия\n");

        Person[] persons = new Person[4];
        
        Employee employee = new Employee("Иван Сидоров", "Разработчик", 85000);
        Manager manager = new Manager("Петр Иванов", "Менеджер отдела", 120000, "IT");
        ProjectManager projectManager = new ProjectManager("Мария Петрова", "Проектный менеджер", 
                                                        150000.00, "Разработка", "Внедрение CRM системы");

        Console.WriteLine("Информация о сотрудниках");
        employee.PrintInfo();
        manager.PrintInfo();
        projectManager.PrintInfo();
        
        string[] teamMembers = {
            "Константин Петров - Старший разработчик",
            "Екатерина Нестерова - QA-инженер", 
            "Денис Волков - UI/UX-дизайнер",
            "Сергей Шорохов - Бизнес-аналитик"
        };

        AdvancedProjectManager advancedPm = new AdvancedProjectManager(
            "Елена Ковалева", 
            "Старший проектный менеджер", 
            180000.25, 
            "Управление проектами",
            "Разработка мобильного приложения",
            teamMembers
        );

        advancedPm.PrintInfo();

        string report = advancedPm.GenerateReport();
        Console.WriteLine(report);
        Console.WriteLine();

        Console.WriteLine("Управление командой");
        advancedPm.DisplayTeam();
        Console.WriteLine();

        persons[0] = employee;
        persons[1] = manager;
        persons[2] = projectManager;
        persons[3] = advancedPm;
        
        foreach (var person in persons)
        {
            person.DisplayBasicInfo();  
        }
        Console.WriteLine();

        IReportable reportable = advancedPm;
        ITeamLeader teamLeader = advancedPm;

        Console.WriteLine("Отчет:");
        Console.WriteLine(reportable.GenerateReport());
        Console.WriteLine();

        Console.WriteLine("Члены команды:");
        string[] members = teamLeader.GetTeamMembers();
        for (int i = 0; i < members.Length; i++)
        {
            Console.WriteLine($" {i + 1}. {members[i]}");
        }

        Console.WriteLine("\nСтатистика по сотрудникам");
        Console.WriteLine($"Общее количество сотрудников: {persons.Length}");
        Console.WriteLine($"Общий фонд заработной платы: {CalculateTotalSalary(persons):N2} руб.");
        
        Console.WriteLine("\nПоиск менеджеров");
        FindManagers(persons);

    public static double CalculateTotalSalary(Person[] persons)
    {
        double total = 0;
        foreach (var person in persons)
        {
            if (person is Employee employee)
            {
                total += employee.Salary;
            }
        }
        return total;
    }

    public static void FindManagers(Person[] persons)
    {
        foreach (var person in persons)
        {
            if (person is Manager manager)
            {
                string department = manager.Department;
                if (manager is AdvancedProjectManager advancedManager)
                {
                    department = advancedManager.Department;
                }
                Console.WriteLine($"Менеджер: {manager.Name}, Отдел: {department}");
            }
        }
    }
```

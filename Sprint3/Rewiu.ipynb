```polyglot-notebook
using System;
using System.Collections.Generic;

public interface IInstallable { void Install(); }

public abstract class Component
{
    public string Model { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }

    public Component(string model, decimal price, string manufacturer)
    {
        Model = model; Price = price; Manufacturer = manufacturer;
    }

    public virtual void GetInfo() => Console.WriteLine($"Производитель: {Manufacturer}, Модель: {Model}, Цена: {Price:C}");
    public override string ToString() => $"{Manufacturer} {Model} - {Price:C}";
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }

    public CPU(string model, decimal price, string manufacturer, int cores, double frequency)
        : base(model, price, manufacturer) { Cores = cores; Frequency = frequency; }

    public override void GetInfo() { base.GetInfo(); Console.WriteLine($"Ядра: {Cores}, Частота: {Frequency}GHz"); }
    public void Install() => Console.WriteLine($"Установка процессора {Model} в сокет");
    
    public static bool operator ==(CPU a, CPU b) => a?.Model == b?.Model && a?.Cores == b?.Cores;
    public static bool operator !=(CPU a, CPU b) => !(a == b);
    public override bool Equals(object obj) => obj is CPU cpu && this == cpu;
    public override int GetHashCode() => HashCode.Combine(Model, Cores);
}

public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public string Type { get; set; }

    public RAM(string model, decimal price, string manufacturer, int capacity, string type)
        : base(model, price, manufacturer) { Capacity = capacity; Type = type; }

    public override void GetInfo() { base.GetInfo(); Console.WriteLine($"Объем: {Capacity}GB, Тип: {Type}"); }
    public void Install() => Console.WriteLine($"Установка памяти {Model} в слот DIMM");
    
    public static bool operator ==(RAM a, RAM b) => a?.Model == b?.Model && a?.Capacity == b?.Capacity;
    public static bool operator !=(RAM a, RAM b) => !(a == b);
    public override bool Equals(object obj) => obj is RAM ram && this == ram;
    public override int GetHashCode() => HashCode.Combine(Model, Capacity);
}

public class GPU : Component, IInstallable
{
    public int VRAM { get; set; }
    public string Interface { get; set; }

    public GPU(string model, decimal price, string manufacturer, int vram, string interfaceType)
        : base(model, price, manufacturer) { VRAM = vram; Interface = interfaceType; }

    public override void GetInfo() { base.GetInfo(); Console.WriteLine($"Видеопамять: {VRAM}GB, Интерфейс: {Interface}"); }
    public void Install() => Console.WriteLine($"Установка видеокарты {Model} в слот PCI-E");
    
    public static bool operator ==(GPU a, GPU b) => a?.Model == b?.Model && a?.VRAM == b?.VRAM;
    public static bool operator !=(GPU a, GPU b) => !(a == b);
    public override bool Equals(object obj) => obj is GPU gpu && this == gpu;
    public override int GetHashCode() => HashCode.Combine(Model, VRAM);
}

public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }
    public int RPM { get; set; }

    public HDD(string model, decimal price, string manufacturer, int capacity, int rpm)
        : base(model, price, manufacturer) { Capacity = capacity; RPM = rpm; }

    public override void GetInfo() { base.GetInfo(); Console.WriteLine($"Объем: {Capacity}GB, Скорость: {RPM}RPM"); }
    public void Install() => Console.WriteLine($"Установка жесткого диска {Model} в отсек");
    
    public static bool operator ==(HDD a, HDD b) => a?.Model == b?.Model && a?.Capacity == b?.Capacity;
    public static bool operator !=(HDD a, HDD b) => !(a == b);
    public override bool Equals(object obj) => obj is HDD hdd && this == hdd;
    public override int GetHashCode() => HashCode.Combine(Model, Capacity);
}

public class ComponentManager<T> where T : Component, IInstallable
{
    private List<T> _components = new List<T>();

    public void AddComponent(T component)
    {
        _components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент {component.Model} добавлен");
    }

    public bool RemoveComponent(T component)
    {
        bool removed = _components.Remove(component);
        if (removed) Console.WriteLine($"Компонент {component.Model} удален");
        return removed;
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var component in _components) total += component.Price;
        return total;
    }

    public void DisplayComponents()
    {
        Console.WriteLine($"\nКомпоненты ({typeof(T).Name}):");
        foreach (var component in _components) component.GetInfo();
        Console.WriteLine($"Общая стоимость: {GetTotalPrice():C}");
    }

    public int ComponentCount => _components.Count;
}

public class ComponentComparer<T> where T : Component
{
    public bool AreEqual(T component1, T component2)
    {
        if (component1 == null || component2 == null)
            return false;

        if (component1 is CPU cpu1 && component2 is CPU cpu2)
            return cpu1 == cpu2;
        else if (component1 is RAM ram1 && component2 is RAM ram2)
            return ram1 == ram2;
        else if (component1 is GPU gpu1 && component2 is GPU gpu2)
            return gpu1 == gpu2;
        else if (component1 is HDD hdd1 && component2 is HDD hdd2)
            return hdd1 == hdd2;
        
        return component1.Model == component2.Model && component1.Price == component2.Price;
    }

    public void CompareAndDisplay(T component1, T component2)
    {
        Console.WriteLine($"Сравнение компонентов:");
        component1.GetInfo();
        component2.GetInfo();
        Console.WriteLine($"Компоненты одинаковы: {AreEqual(component1, component2)}");
    }
}

public class ComputerBuilder<T> where T : Component, IInstallable
{
    private ComponentManager<T> _componentManager = new ComponentManager<T>();

    public void AddComponent(T component)
    {
        _componentManager.AddComponent(component);
    }

    public void RemoveComponent(T component)
    {
        _componentManager.RemoveComponent(component);
    }

    public void DisplayBuild()
    {
        _componentManager.DisplayComponents();
    }

    public decimal GetBuildPrice()
    {
        return _componentManager.GetTotalPrice();
    }
}

public class Computer
{
    private List<Component> _components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент {component.Model} добавлен");
    }

    public bool RemoveComponent<T>(T component) where T : Component
    {
        bool removed = _components.Remove(component);
        if (removed) Console.WriteLine($"Компонент {component.Model} удален");
        return removed;
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var component in _components) total += component.Price;
        return total;
    }

    public void DisplayComponents()
    {
        Console.WriteLine("\nКомпоненты компьютера:");
        foreach (var component in _components) component.GetInfo();
        Console.WriteLine($"Общая стоимость: {GetTotalPrice():C}");
    }
}

public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        var computer = new Computer();
        computer.AddComponent(new CPU("Ryzen 7 5800X", 30000, "AMD", 8, 3.8));
        computer.AddComponent(new GPU("RTX 3080", 70000, "NVIDIA", 10, "PCI-E 4.0"));
        computer.AddComponent(new RAM("Vengeance LPX", 8000, "Corsair", 16, "DDR4"));
        computer.AddComponent(new HDD("Barracuda", 5000, "Seagate", 2000, 7200));
        return computer;
    }

    public static Computer CreateOfficePC()
    {
        var computer = new Computer();
        computer.AddComponent(new CPU("Core i5-12400", 15000, "Intel", 6, 2.5));
        computer.AddComponent(new GPU("UHD Graphics 730", 0, "Intel", 0, "Встроенная"));
        computer.AddComponent(new RAM("ValueRAM", 4000, "Kingston", 8, "DDR4"));
        computer.AddComponent(new HDD("Blue", 3000, "WD", 1000, 5400));
        return computer;
    }
}

        var gamingPC = ComputerFactory.CreateGamingPC();
        gamingPC.DisplayComponents();

        Console.WriteLine("\n" );
        
        var officePC = ComputerFactory.CreateOfficePC();
        officePC.DisplayComponents();

        Console.WriteLine("\nСоздание и сравнение компонентов:");
        var cpu1 = new CPU("Ryzen 7 5800X", 30000, "AMD", 8, 3.8);
        var cpu2 = new CPU("Ryzen 7 5800X", 30000, "AMD", 8, 3.8);
        
        var cpuComparer = new ComponentComparer<CPU>();
        cpuComparer.CompareAndDisplay(cpu1, cpu2);
        Console.WriteLine($"CPU1 == CPU2: {cpuComparer.AreEqual(cpu1, cpu2)}");
        Console.WriteLine($"Информация: {cpu1}");

        Console.WriteLine("\nДобавление и удаление:");
        
        var cpuManager = new ComponentManager<CPU>();
        cpuManager.AddComponent(cpu1);
        cpuManager.DisplayComponents();
        
        var ramManager = new ComponentManager<RAM>();
        var ram = new RAM("Vengeance LPX", 8000, "Corsair", 16, "DDR4");
        ramManager.AddComponent(ram);
        ramManager.DisplayComponents();
        
        var computerBuilder = new ComputerBuilder<CPU>();
        computerBuilder.AddComponent(new CPU("Core i9", 45000, "Intel", 12, 4.5));
        computerBuilder.DisplayBuild();

        var customPC = new Computer();
        customPC.AddComponent(ram);
        customPC.RemoveComponent(ram);
```

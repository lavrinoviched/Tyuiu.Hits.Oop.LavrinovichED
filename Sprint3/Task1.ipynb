### Перекрытие методов в C#

Перекрытие методов (или переопределение) — это процесс, при котором производный класс предоставляет новую реализацию метода, уже существующего в его базовом классе. Это позволяет изменить или расширить поведение метода в производном классе без изменения исходного метода в базовом классе. Перекрытие методов является ключевой особенностью полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример с классами `Vehicle`, `Car` и `Motorcycle`, где каждый из этих классов имеет метод `Drive()`, который принимает параметр `speed`.

```csharp
// Базовый класс Vehicle с виртуальным методом Drive()
public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }
}

// Производный класс Car, переопределяющий метод Drive()
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Автомобиль движется {speed} km/h.");
    }
}

// Производный класс Motorcycle, также переопределяющий метод Drive()
public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        Console.WriteLine($"Мотоцикл движется {speed} km/h.");
    }
}
```

В этом примере, метод `Drive()` в классе `Car` и `Motorcycle` переопределяет метод `Drive()` из базового класса `Vehicle`. Это означает, что при вызове метода `Drive()` на объекте типа `Car` или `Motorcycle`, будет вызвана соответствующая реализация метода в этих классах, а не в классе `Vehicle`.

#### Отличие перекрытия от перегрузки методов

Перегрузка методов — это процесс определения нескольких методов с одним и тем же именем, но с разным количеством или типом параметров. Например:

```csharp
public class ExampleClass
{
    public void Print(int number)
    {
        Console.WriteLine(number);
    }

    public void Print(double number)
    {
        Console.WriteLine(number.ToString());
    }
}
```

В этом случае, у нас есть два метода с именем `Print`, но с разным типом параметра. Это отличается от перекрытия методов, где метод в производном классе полностью заменяет метод в базовом классе.

#### Вызов методов родителя

Иногда в производном классе может потребоваться вызвать метод родителя. Для этого используется ключевое слово `base`. Например, если мы хотим добавить дополнительную логику в метод `Drive()` класса `Car`, сохраняя при этом оригинальное поведение метода `Drive()` из класса `Vehicle`.

```csharp
public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        base.Drive(speed); // Вызов метода Drive() из класса Vehicle
        Console.WriteLine("Добавляем специфическое поведение автомобиля");
    }
}
```

В этом примере, после вызова `base.Drive(speed)`, выполняется дополнительная логика, специфичная для класса `Car`.


### Перегрузка методов в C#

Перегрузка методов — это способность класса определить несколько методов с одним и тем же именем, но с разным количеством или типом параметров. Это позволяет разработчикам предоставлять различные варианты выполнения одного и того же действия в зависимости от входных данных. Перегрузка методов является одной из форм выражения полиморфизма в объектно-ориентированном программировании.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример, где класс `Vehicle` содержит перегруженные методы `Drive()`, которые принимают разные типы параметров.

```csharp
public class Vehicle
{
    // Метод Drive(), принимающий int
    public void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется {speed} km/h.");
    }

    // Метод Drive(), принимающий string
    public void Drive(string mode)
    {
        Console.WriteLine($"Автомобиль движется в {mode} режиме.");
    }

    // Метод Drive(), принимающий bool
    public void Drive(bool isManual)
    {
        if (isManual)
        {
            Console.WriteLine("Автомобиль переключает передачи вручную.");
        }
        else
        {
            Console.WriteLine("Автомобиль оснащен автоматической коробкой передач.");
        }
    }
}
```

В этом примере класс `Vehicle` определяет три метода с именем `Drive()`, каждый из которых принимает разные типы параметров: `int`, `string` и `bool`. Это позволяет вызывать разные версии метода `Drive()` в зависимости от контекста, например:

```csharp
Vehicle myVehicle = new Vehicle();

myVehicle.Drive(60); // Вызов метода Drive() с int параметром
myVehicle.Drive("Eco"); // Вызов метода Drive() с string параметром
myVehicle.Drive(true); // Вызов метода Drive() с bool параметром
```

#### Отличие перегрузки от переопределения

Отличие перегрузки методов от переопределения заключается в том, что перегрузка методов включает определение нескольких методов с одним именем, но с разным набором параметров, тогда как переопределение методов (или перекрытие) означает предоставление новой реализации метода в производном классе, который уже существует в базовом классе.

#### Вызов методов

При вызове перегруженного метода компилятор выбирает подходящую версию метода на основе типов и количества аргументов, переданных в вызове. Если типы и количество аргументов совпадают с несколькими версиями метода, компилятор выберет ту версию, которая была объявлена первой.

### Заключение

Перегрузка методов в C# позволяет классам предоставлять различные варианты выполнения одного и того же действия, адаптируясь к различным условиям использования. Это мощный инструмент для создания гибких и удобных API, позволяющих клиентам класса использовать его функциональность наиболее эффективно.

Перекрытие методов в C# позволяет производным классам изменять или расширять поведение методов базовых классов, что является ключевой особенностью полиморфизма. Это отличается от перегрузки методов, которая предполагает определение нескольких методов с одинаковым именем, но с разным набором параметров.

````
using System;

public class Vehicle
{
    public virtual void Drive(int speed)
    {
        Console.WriteLine($"Транспорт движется со скоростью {speed} км/ч.");
    }
    
    public void Drive(string mode)
    {
        Console.WriteLine($"Транспорт движется в режиме: {mode}.");
    }
    
    public void Drive(int speed, string destination)
    {
        Console.WriteLine($"Транспорт движется со скоростью {speed} км/ч к пункту назначения: {destination}.");
    }
    
    public void Drive(bool isEcoMode)
    {
        string mode = isEcoMode ? "эко-режиме" : "обычном режиме";
        Console.WriteLine($"Транспорт движется в {mode}.");
    }
}

public class Car : Vehicle
{
    public override void Drive(int speed)
    {
        if (speed > 120)
        {
            Console.WriteLine($"Автомобиль движется очень быстро: {speed} км/ч! Соблюдайте осторожность.");
        }
        else
        {
            Console.WriteLine($"Автомобиль движется со скоростью {speed} км/ч.");
        }
    }
    
    public void DriveWithBaseCall(int speed)
    {
        base.Drive(speed); 
        Console.WriteLine("Дополнительная информация: автомобиль в хорошем состоянии.");
    }
    
    public void Drive(string brand, int speed)
    {
        Console.WriteLine($"Автомобиль марки {brand} движется со скоростью {speed} км/ч.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Drive(int speed)
    {
        if (speed > 100)
        {
            Console.WriteLine($"Мотоцикл мчится на скорости {speed} км/ч! Будьте осторожны!");
        }
        else
        {
            Console.WriteLine($"Мотоцикл едет со скоростью {speed} км/ч.");
        }
    }
    
    public void Drive(bool withPassenger, int speed)
    {
        string passengerInfo = withPassenger ? "с пассажиром" : "без пассажира";
        Console.WriteLine($"Мотоцикл движется {passengerInfo} со скоростью {speed} км/ч.");
    }
}

public class Truck : Vehicle
{
    public int LoadCapacity { get; set; }
    
    public Truck(int capacity)
    {
        LoadCapacity = capacity;
    }
    
    public override void Drive(int speed)
    {
        if (speed > 80)
        {
            Console.WriteLine($"Грузовик с грузоподъемностью {LoadCapacity}т движется слишком быстро: {speed} км/ч!");
        }
        else
        {
            base.Drive(speed);
            Console.WriteLine($"Грузоподъемность: {LoadCapacity} тонн.");
        }
    }
    
    public void Drive(int speed, int currentLoad)
    {
        if (currentLoad > LoadCapacity)
        {
            Console.WriteLine($"ВНИМАНИЕ! Перегруз! Грузовик движется со скоростью {speed} км/ч с грузом {currentLoad}т при максимальной {LoadCapacity}т.");
        }
        else
        {
            Console.WriteLine($"Грузовик движется со скоростью {speed} км/ч с грузом {currentLoad}т.");
        }
    }
}

        Vehicle vehicle = new Vehicle();
        Car car = new Car();
        Motorcycle motorcycle = new Motorcycle();
        Truck truck = new Truck(10);
        
        vehicle.Drive(60);
        vehicle.Drive("Городской");
        vehicle.Drive(70, "Москва");
        vehicle.Drive(true);

        Vehicle[] vehicles = { new Vehicle(), new Car(), new Motorcycle(), new Truck(15) };
        
        foreach (var v in vehicles)
        {
            v.Drive(90); 
        }

        car.Drive(130);        
        motorcycle.Drive(110); 
        truck.Drive(85);      

        car.DriveWithBaseCall(80);

        car.Drive("Toyota", 100);
        motorcycle.Drive(true, 80);
        truck.Drive(60, 12); 

        TestOverloading();
    
    static void TestOverloading()
    {
        Car testCar = new Car();
        
        testCar.Drive(100);                  
        testCar.Drive("Городской");        
        testCar.Drive("BMW", 120);        
        testCar.Drive(90, "Санкт-Петербург"); 
    }
````

````
using System;
using System.Collections.Generic;

public class Animal
{
    private string _name;
    private int _age;
    private double _weight;

    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }
    
    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }
    
    public virtual string Species { get; } = "Неизвестный вид";
    
    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
        Weight = 1.0;
    }
    
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }
    
    public virtual void Move()
    {
        Console.WriteLine($"{Name} перемещается.");
    }

    public void Move(int distance)
    {
        Console.WriteLine($"{Name} перемещается на расстояние {distance} метров.");
    }
    
    public void Move(string direction)
    {
        Console.WriteLine($"{Name} движется в направлении: {direction}.");
    }
    
    public virtual void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}.");
    }

    public void Eat(string food, int quantity)
    {
        Console.WriteLine($"{Name} ест {quantity} порций {food}.");
    }
    
    public void Eat(bool isHungry)
    {
        if (isHungry)
            Console.WriteLine($"{Name} голоден и ищет еду.");
        else
            Console.WriteLine($"{Name} не голоден.");
    }

    public virtual string GetInfo()
    {
        return $"{Species} {Name}, возраст: {Age} лет, вес: {Weight} кг";
    }

    public string GetInfo(bool detailed)
    {
        if (detailed)
            return $"{Species} {Name}, возраст: {Age} лет, вес: {Weight} кг, состояние: здоров";
        else
            return GetInfo();
    }
}

public class Dog : Animal
{
    public override string Species { get; } = "Собака";
    
    public string Breed { get; set; }
    public bool IsTrained { get; set; }

    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
        IsTrained = false;
    }
    
    public Dog(string name, int age, string breed, double weight, bool isTrained) : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} бегает и виляет хвостом.");
    }

    public void Move(bool isPlaying)
    {
        if (isPlaying)
            Console.WriteLine($"{Name} весело бегает и прыгает во время игры!");
        else
            Move();
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} с удовольствием ест {food}.");
    }

    public void Fetch(string item)
    {
        Console.WriteLine($"{Name} приносит {item}.");
    }

    public void Fetch(string item, int times)
    {
        Console.WriteLine($"{Name} приносит {item} {times} раз подряд.");
    }
    
    public override string GetInfo()
    {
        string trained = IsTrained ? "дрессированная" : "недрессированная";
        return $"{base.GetInfo()}, порода: {Breed}, {trained}";
    }
}

public class Cat : Animal
{
    public override string Species { get; } = "Кошка";
    
    public bool IsIndoor { get; set; }
    public int Lives { get; set; }
    
    public Cat(string name, int age, bool isIndoor = true) : base(name, age)
    {
        IsIndoor = isIndoor;
        Lives = 9;
    }
    
    public Cat(string name, int age, double weight, bool isIndoor, int lives) : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        Lives = lives;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} грациозно ходит и прыгает.");
    }
    
    public void Move(bool isHunting)
    {
        if (isHunting)
            Console.WriteLine($"{Name} крадется и готовится к прыжку!");
        else
            Move();
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} осторожно пробует {food}.");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }

    public void Purr(int volume)
    {
        string sound = volume > 5 ? "громко мурлычет" : "тихо мурлычет";
        Console.WriteLine($"{Name} {sound}: МРРРРР...");
    }
    
    public override string GetInfo()
    {
        string living = IsIndoor ? "домашняя" : "уличная";
        return $"{base.GetInfo()}, {living}, жизней: {Lives}";
    }
}

public class Bird : Animal
{
    public override string Species { get; } = "Птица";
    
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, double wingspan, bool canFly = true) : base(name, age)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }
    
    public Bird(string name, int age, double weight, double wingspan, bool canFly) : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летает в небе.");
        else
            Console.WriteLine($"{Name} прыгает по земле.");
    }

    public void Move(int altitude, string direction)
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит на высоте {altitude} метров в направлении {direction}.");
        else
            Move();
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} клюет {food}.");
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит с размахом крыльев {Wingspan} см.");
        else
            Console.WriteLine($"{Name} не умеет летать.");
    }

    public void Fly(int distance)
    {
        if (CanFly)
            Console.WriteLine($"{Name} пролетает расстояние {distance} метров.");
        else
            Console.WriteLine($"{Name} не умеет летать.");
    }
    
    public override string GetInfo()
    {
        string flight = CanFly ? "летающая" : "нелетающая";
        return $"{base.GetInfo()}, размах крыльев: {Wingspan} см, {flight}";
    }
}

public class Fish : Animal
{
    public override string Species { get; } = "Рыба";
    
    public string WaterType { get; set; }
    public bool HasScales { get; set; }
    
    public Fish(string name, int age, string waterType = "пресная") : base(name, age)
    {
        WaterType = waterType;
        HasScales = true;
    }
    
    public Fish(string name, int age, double weight, string waterType, bool hasScales) : base(name, age, weight)
    {
        WaterType = waterType;
        HasScales = hasScales;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} плавает в воде.");
    }
    
    public void Move(int depth, bool isDiving)
    {
        string action = isDiving ? "погружается" : "всплывает";
        Console.WriteLine($"{Name} {action} на глубину {depth} метров.");
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} медленно ест {food} в воде.");
    }
    
    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде.");
    }
    
    public void Swim(string style)
    {
        Console.WriteLine($"{Name} плавает в стиле '{style}' в {WaterType} воде.");
    }
    
    public override string GetInfo()
    {
        string scales = HasScales ? "чешуйчатая" : "без чешуи";
        return $"{base.GetInfo()}, тип воды: {WaterType}, {scales}";
    }
}

        
        List<Animal> animals = new List<Animal>
        {
            new Dog("Барсик", 3, "Лабрадор", 25.5, true),
            new Cat("Мурка", 2, 4.2, true, 9),
            new Bird("Кеша", 1, 0.3, 15.5, true),
            new Fish("Немо", 1, 0.5, "соленая", true),
            new Bird("Пингвин", 4, 30.0, 35.0, false)
        };
        
        Console.WriteLine("1. ");
        foreach (var animal in animals)
        {
            animal.MakeSound(); 
        }
        
        Console.WriteLine("\n2. ");
        foreach (var animal in animals)
        {
            animal.Move(); 
        }
        
        Console.WriteLine("\n3. ");
        animals[0].Move(100); 
        animals[0].Move("север"); 
        
        Console.WriteLine("\n4. ");
        animals[1].Eat("рыбу"); 
        animals[1].Eat("рыбу", 2); 
        animals[1].Eat(true); 
        
        Console.WriteLine("\n5. ");
        Cat cat = (Cat)animals[1];
        Bird bird = (Bird)animals[2];
        Fish fish = (Fish)animals[3];
        
        dog.Move(true); 
        cat.Move(true); 
        bird.Move(100, "юг"); 
        fish.Move(5, true); 
        
        Console.WriteLine("\n6. ");
        dog.Fetch("палку");
        dog.Fetch("мяч", 3);
        cat.Purr();
        cat.Purr(8);
        bird.Fly();
        bird.Fly(500);
        fish.Swim();
        fish.Swim("брасс");
        
        Console.WriteLine("\n7. ");
        Console.WriteLine(dog.GetInfo());
        Console.WriteLine(dog.GetInfo(true));
        Console.WriteLine(cat.GetInfo());
        Console.WriteLine(bird.GetInfo());
        Console.WriteLine(fish.GetInfo());
````

<h1 style="color:DodgerBlue">Generic классы в C#</h1>


Генерические классы в C# представляют собой шаблон класса, который может работать с различными типами данных без потери производительности. Они позволяют создавать компоненты, которые могут быть безопасно повторно использованы в различных контекстах, обеспечивая при этом строгую типизацию и избегая необходимости приведения типов.

#### Основные преимущества генерических классов:

1. **Безопасность типов**: Генерические классы обеспечивают проверку типов на этапе компиляции, что помогает предотвратить ошибки во время выполнения.
2. **Универсальность**: Генерические классы могут работать с различными типами данных, делая их очень гибкими и универсальными.
3. **Производительность**: Генерические классы не приводят к значительной потере производительности, поскольку информация о типе хранится вместе с объектом.

#### Пример с классами `Vehicle`, `Car` и `Motorcycle`

Давайте рассмотрим пример генерического класса `VehicleCollection<T>`, который может хранить коллекцию объектов типа `T`, где `T` может быть любой строкой, например, `Vehicle`, `Car` или `Motorcycle`.

В этом примере `VehicleCollection<T>` является генерическим классом, который может хранить коллекцию объектов типа `T`, где `T` должен быть подклассом `Vehicle`. Это позволяет создавать коллекции автомобилей, мотоциклов и других видов транспортных средств, обеспечивая при этом строгую типизацию и безопасность типов.

````
using System;
using System.Collections.Generic;

public class Vehicle
{
    public string Brand { get; set; }
    public int Year { get; set; }
    
    public Vehicle(string brand, int year)
    {
        Brand = brand;
        Year = year;
    }
    
    public virtual void Drive()
    {
        Console.WriteLine($"Транспортное средство {Brand} находится в движении.");
    }
    
    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Транспорт: {Brand}, Год выпуска: {Year}");
    }
}

public class Car : Vehicle
{
    public int Doors { get; set; }
    
    public Car(string brand, int year, int doors) : base(brand, year)
    {
        Doors = doors;
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Автомобиль {Brand} едет по дороге.");
    }
    
    public override void DisplayInfo()
    {
        Console.WriteLine($"Автомобиль: {Brand}, Год: {Year}, Дверей: {Doors}");
    }
}

public class Motorcycle : Vehicle
{
    public string Type { get; set; }
    
    public Motorcycle(string brand, int year, string type) : base(brand, year)
    {
        Type = type;
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Мотоцикл {Brand} мчится по трассе.");
    }
    
    public override void DisplayInfo()
    {
        Console.WriteLine($"Мотоцикл: {Brand}, Год: {Year}, Тип: {Type}");
    }
}

public class Truck : Vehicle
{
    public double LoadCapacity { get; set; }
    
    public Truck(string brand, int year, double loadCapacity) : base(brand, year)
    {
        LoadCapacity = loadCapacity;
    }
    
    public override void Drive()
    {
        Console.WriteLine($"Грузовик {Brand} перевозит груз.");
    }
    
    public override void DisplayInfo()
    {
        Console.WriteLine($"Грузовик: {Brand}, Год: {Year}, Грузоподъемность: {LoadCapacity}т");
    }
}

public class VehicleCollection<T> where T : Vehicle
{
    private List<T> _vehicles = new List<T>();
    
    public void Add(T vehicle)
    {
        _vehicles.Add(vehicle);
        Console.WriteLine($"Добавлено: {vehicle.Brand}");
    }
    
    public bool Remove(T vehicle)
    {
        bool removed = _vehicles.Remove(vehicle);
        if (removed)
            Console.WriteLine($"Удалено: {vehicle.Brand}");
        return removed;
    }
    
    public void DisplayAll()
    {
        Console.WriteLine($"\n Коллекция транспортных средств ({typeof(T).Name}) ");
        foreach (var vehicle in _vehicles)
        {
            vehicle.DisplayInfo();
        }
    }
    
    public void DriveAll()
    {
        Console.WriteLine($"\n Движение транспортных средств ({typeof(T).Name}) ");
        foreach (var vehicle in _vehicles)
        {
            vehicle.Drive();
        }
    }
    
    public T FindByBrand(string brand)
    {
        return _vehicles.Find(v => v.Brand.Equals(brand, StringComparison.OrdinalIgnoreCase));
    }
    
    public int Count => _vehicles.Count;
    
    public T this[int index]
    {
        get
        {
            if (index >= 0 && index < _vehicles.Count)
                return _vehicles[index];
            throw new IndexOutOfRangeException();
        }
        set
        {
            if (index >= 0 && index < _vehicles.Count)
                _vehicles[index] = value;
            else
                throw new IndexOutOfRangeException();
        }
    }
}

public class VehicleService<T> where T : Vehicle
{
    private List<T> _servicedVehicles = new List<T>();
    
    public void ServiceVehicle(T vehicle)
    {
        _servicedVehicles.Add(vehicle);
        Console.WriteLine($"Обслужено транспортное средство: {vehicle.Brand}");
    }
    
    public void DisplayServicedVehicles()
    {
        Console.WriteLine($"\n Обслуженные транспортные средства ({typeof(T).Name}) ");
        foreach (var vehicle in _servicedVehicles)
        {
            vehicle.DisplayInfo();
        }
    }
}

public class Garage<T> where T : Vehicle, new()
{
    private List<T> _parkedVehicles = new List<T>();
    
    public void ParkVehicle()
    {
        T vehicle = new T();
        _parkedVehicles.Add(vehicle);
        Console.WriteLine($"Припарковано транспортное средство по умолчанию: {vehicle.Brand}");
    }
    
    public void ParkVehicle(T vehicle)
    {
        _parkedVehicles.Add(vehicle);
        Console.WriteLine($"Припарковано: {vehicle.Brand}");
    }
    
    public void DisplayParkedVehicles()
    {
        Console.WriteLine($"\n Припаркованные транспортные средства ({typeof(T).Name}) ");
        foreach (var vehicle in _parkedVehicles)
        {
            vehicle.DisplayInfo();
        }
    }
}



        Car car1 = new Car("Toyota", 2020, 4);
        Car car2 = new Car("Honda", 2022, 2);
        Motorcycle motorcycle1 = new Motorcycle("Yamaha", 2021, "Спортивный");
        Motorcycle motorcycle2 = new Motorcycle("Harley-Davidson", 2020, "Круизер");
        Truck truck1 = new Truck("Volvo", 2019, 15.5);
        Truck truck2 = new Truck("MAN", 2023, 20.0);
        
        Console.WriteLine("1. Коллекция автомобилей:");
        VehicleCollection<Car> carCollection = new VehicleCollection<Car>();
        carCollection.Add(car1);
        carCollection.Add(car2);
        carCollection.DisplayAll();
        carCollection.DriveAll();
        Console.WriteLine($"Количество автомобилей: {carCollection.Count}");
        
        Console.WriteLine("\n2. Коллекция мотоциклов:");
        VehicleCollection<Motorcycle> motorcycleCollection = new VehicleCollection<Motorcycle>();
        motorcycleCollection.Add(motorcycle1);
        motorcycleCollection.Add(motorcycle2);
        motorcycleCollection.DisplayAll();
        motorcycleCollection.DriveAll();
        
        Console.WriteLine("\n3. Коллекция грузовиков:");
        VehicleCollection<Truck> truckCollection = new VehicleCollection<Truck>();
        truckCollection.Add(truck1);
        truckCollection.Add(truck2);
        truckCollection.DisplayAll();
        truckCollection.DriveAll();
        
        Console.WriteLine("\n4. Общая коллекция транспортных средств:");
        VehicleCollection<Vehicle> generalCollection = new VehicleCollection<Vehicle>();
        generalCollection.Add(car1);
        generalCollection.Add(motorcycle1);
        generalCollection.Add(truck1);
        generalCollection.DisplayAll();
        generalCollection.DriveAll();
        
        Console.WriteLine("\n5. Сервис обслуживания:");
        VehicleService<Car> carService = new VehicleService<Car>();
        carService.ServiceVehicle(car1);
        carService.ServiceVehicle(car2);
        carService.DisplayServicedVehicles();
        
        Console.WriteLine("\n6. Работа с индексатором:");
        Console.WriteLine($"Первый автомобиль в коллекции: {carCollection[0].Brand}");
        
        Console.WriteLine("\n7. Поиск транспортных средств:");
        Car foundCar = carCollection.FindByBrand("Toyota");
        if (foundCar != null)
        {
            Console.WriteLine("Найден автомобиль:");
            foundCar.DisplayInfo();
        }
        
        Console.WriteLine("\n8. Удаление из коллекции:");
        carCollection.Remove(car1);
        carCollection.DisplayAll();
        Console.WriteLine($"Количество автомобилей после удаления: {carCollection.Count}");
        
        Console.WriteLine("\n ");
        List<VehicleCollection<Vehicle>> allCollections = new List<VehicleCollection<Vehicle>>
        {
            new VehicleCollection<Vehicle>(),
            new VehicleCollection<Vehicle>()
        };
        
        allCollections[0].Add(new Car("BMW", 2023, 4));
        allCollections[0].Add(new Motorcycle("Kawasaki", 2022, "Туристический"));
        allCollections[1].Add(new Truck("Scania", 2021, 18.0));
        
        foreach (var collection in allCollections)
        {
            collection.DriveAll();
        }
````

````
using System;
using System.Collections.Generic;

public class Animal
{
    private string _name;
    private int _age;
    private double _weight;
    
    public string Name
    {
        get { return _name; }
        set 
        { 
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentException("Возраст не может быть отрицательным");
        }
    }
    
    public double Weight
    {
        get { return _weight; }
        set 
        { 
            if (value > 0)
                _weight = value;
            else
                throw new ArgumentException("Вес должен быть положительным");
        }
    }
    
    public virtual string Species { get; } = "Неизвестный вид";
    
    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
        Weight = 1.0;
    }
    
    public Animal(string name, int age, double weight)
    {
        Name = name;
        Age = age;
        Weight = weight;
    }
    
    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }
    
    public virtual void Move()
    {
        Console.WriteLine($"{Name} перемещается.");
    }
    
    public virtual void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}.");
    }
    
    public virtual string GetInfo()
    {
        return $"{Species} {Name}, возраст: {Age} лет, вес: {Weight} кг";
    }
}

public class Dog : Animal
{
    public override string Species { get; } = "Собака";
    
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    
    public Dog(string name, int age, string breed) : base(name, age)
    {
        Breed = breed;
        IsTrained = false;
    }
    
    public Dog(string name, int age, string breed, double weight, bool isTrained) : base(name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} гавкает: Гав-гав!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} бегает и виляет хвостом.");
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} с удовольствием ест {food}.");
    }
    
    public void Fetch(string item)
    {
        Console.WriteLine($"{Name} приносит {item}.");
    }
    
    public override string GetInfo()
    {
        string trained = IsTrained ? "дрессированная" : "недрессированная";
        return $"{base.GetInfo()}, порода: {Breed}, {trained}";
    }
}

public class Cat : Animal
{
    public override string Species { get; } = "Кошка";
    
    public bool IsIndoor { get; set; }
    public int Lives { get; set; }
    
    public Cat(string name, int age, bool isIndoor = true) : base(name, age)
    {
        IsIndoor = isIndoor;
        Lives = 9;
    }
    
    public Cat(string name, int age, double weight, bool isIndoor, int lives) : base(name, age, weight)
    {
        IsIndoor = isIndoor;
        Lives = lives;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} грациозно ходит и прыгает.");
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} осторожно пробует {food}.");
    }
    
    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мрррр...");
    }
    
    public override string GetInfo()
    {
        string living = IsIndoor ? "домашняя" : "уличная";
        return $"{base.GetInfo()}, {living}, жизней: {Lives}";
    }
}

public class Bird : Animal
{
    public override string Species { get; } = "Птица";
    
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, double wingspan, bool canFly = true) : base(name, age)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }
    
    public Bird(string name, int age, double weight, double wingspan, bool canFly) : base(name, age, weight)
    {
        Wingspan = wingspan;
        CanFly = canFly;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }
    
    public override void Move()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летает в небе.");
        else
            Console.WriteLine($"{Name} прыгает по земле.");
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} клюет {food}.");
    }
    
    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летит с размахом крыльев {Wingspan} см.");
        else
            Console.WriteLine($"{Name} не умеет летать.");
    }
    
    public override string GetInfo()
    {
        string flight = CanFly ? "летающая" : "нелетающая";
        return $"{base.GetInfo()}, размах крыльев: {Wingspan} см, {flight}";
    }
}

public class Fish : Animal
{
    public override string Species { get; } = "Рыба";
    
    public string WaterType { get; set; }
    public bool HasScales { get; set; }
    
    public Fish(string name, int age, string waterType = "пресная") : base(name, age)
    {
        WaterType = waterType;
        HasScales = true;
    }
    
    public Fish(string name, int age, double weight, string waterType, bool hasScales) : base(name, age, weight)
    {
        WaterType = waterType;
        HasScales = hasScales;
    }
    
    public override void MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }
    
    public override void Move()
    {
        Console.WriteLine($"{Name} плавает в воде.");
    }
    
    public override void Eat(string food)
    {
        Console.WriteLine($"{Name} медленно ест {food} в воде.");
    }
    
    public void Swim()
    {
        Console.WriteLine($"{Name} плавает в {WaterType} воде.");
    }
    
    public override string GetInfo()
    {
        string scales = HasScales ? "чешуйчатая" : "без чешуи";
        return $"{base.GetInfo()}, тип воды: {WaterType}, {scales}";
    }
}

public class AnimalCollection<T> where T : Animal
{
    private List<T> _animals = new List<T>();
    
    public void Add(T animal)
    {
        _animals.Add(animal);
        Console.WriteLine($"Добавлено: {animal.Name} ({animal.Species})");
    }
    
    public bool Remove(T animal)
    {
        bool removed = _animals.Remove(animal);
        if (removed)
            Console.WriteLine($"Удалено: {animal.Name}");
        return removed;
    }
    
    public void DisplayAll()
    {
        Console.WriteLine($"\n Коллекция животных ({typeof(T).Name}) ");
        foreach (var animal in _animals)
        {
            Console.WriteLine(animal.GetInfo());
        }
    }
    
    public void MakeAllSounds()
    {
        Console.WriteLine($"\n Звуки животных ({typeof(T).Name}) ");
        foreach (var animal in _animals)
        {
            animal.MakeSound();
        }
    }
    
    public T FindByName(string name)
    {
        return _animals.Find(a => a.Name.Equals(name, StringComparison.OrdinalIgnoreCase));
    }
    
    public int Count => _animals.Count;
    
    public T this[int index]
    {
        get
        {
            if (index >= 0 && index < _animals.Count)
                return _animals[index];
            throw new IndexOutOfRangeException();
        }
    }
}

public class VeterinaryClinic<T> where T : Animal
{
    private List<T> _patients = new List<T>();
    
    public void AdmitPatient(T patient)
    {
        _patients.Add(patient);
        Console.WriteLine($"Принят пациент: {patient.Name} ({patient.Species})");
    }
    
    public void DischargePatient(T patient)
    {
        if (_patients.Remove(patient))
            Console.WriteLine($"Выписан пациент: {patient.Name}");
    }
    
    public void PerformCheckup()
    {
        Console.WriteLine($"\n Медосмотр пациентов ({typeof(T).Name}) ");
        foreach (var patient in _patients)
        {
            Console.WriteLine($"Осмотр: {patient.Name}, Вес: {patient.Weight} кг, Возраст: {patient.Age} лет");
        }
    }
    
    public void FeedAll(string food)
    {
        Console.WriteLine($"\n Кормление пациентов ({typeof(T).Name}) ");
        foreach (var patient in _patients)
        {
            patient.Eat(food);
        }
    }
}

public class ZooEnclosure<T> where T : Animal
{
    private List<T> _inhabitants = new List<T>();
    
    public void AddInhabitant(T animal)
    {
        _inhabitants.Add(animal);
        Console.WriteLine($"Добавлен обитатель: {animal.Name} ({typeof(T).Name})");
    }
    
    public void DisplayInhabitants()
    {
        Console.WriteLine($"\n Обитатели вольера ({typeof(T).Name}) ");
        foreach (var inhabitant in _inhabitants)
        {
            Console.WriteLine(inhabitant.GetInfo());
        }
    }
}

public class AnimalRegistry<T> where T : Animal
{
    private Dictionary<string, T> _registry = new Dictionary<string, T>();
    
    public void RegisterAnimal(T animal)
    {
        _registry[animal.Name] = animal;
        Console.WriteLine($"Зарегистрировано: {animal.Name} ({animal.Species})");
    }
    
    public T GetAnimal(string name)
    {
        if (_registry.ContainsKey(name))
            return _registry[name];
        return null;
    }
    
    public void DisplayRegistry()
    {
        Console.WriteLine($"\n Реестр животных ({typeof(T).Name}) ");
        foreach (var animal in _registry.Values)
        {
            Console.WriteLine(animal.GetInfo());
        }
    }
}


        Dog dog1 = new Dog("Барсик", 3, "Лабрадор", 25.5, true);
        Dog dog2 = new Dog("Шарик", 5, "Овчарка", 30.0, false);
        Cat cat1 = new Cat("Мурка", 2, 4.2, true, 9);
        Cat cat2 = new Cat("Васька", 4, 5.1, false, 7);
        Bird bird1 = new Bird("Кеша", 1, 15.5, true);
        Bird bird2 = new Bird("Пингвин", 4, 35.0, false);
        Fish fish1 = new Fish("Немо", 1, "соленая");
        Fish fish2 = new Fish("Золотик", 2, "пресная");

        Console.WriteLine("1. Генерические коллекции животных:");
        
        AnimalCollection<Dog> dogCollection = new AnimalCollection<Dog>();
        dogCollection.Add(dog1);
        dogCollection.Add(dog2);
        dogCollection.DisplayAll();
        dogCollection.MakeAllSounds();
        
        AnimalCollection<Cat> catCollection = new AnimalCollection<Cat>();
        catCollection.Add(cat1);
        catCollection.Add(cat2);
        catCollection.DisplayAll();
        catCollection.MakeAllSounds();
        
        AnimalCollection<Bird> birdCollection = new AnimalCollection<Bird>();
        birdCollection.Add(bird1);
        birdCollection.Add(bird2);
        birdCollection.DisplayAll();
        birdCollection.MakeAllSounds();

        Console.WriteLine("\n2. Ветеринарная клиника:");
        VeterinaryClinic<Dog> dogClinic = new VeterinaryClinic<Dog>();
        dogClinic.AdmitPatient(dog1);
        dogClinic.AdmitPatient(dog2);
        dogClinic.PerformCheckup();
        dogClinic.FeedAll("мясо");
        
        VeterinaryClinic<Cat> catClinic = new VeterinaryClinic<Cat>();
        catClinic.AdmitPatient(cat1);
        catClinic.PerformCheckup();
        catClinic.FeedAll("рыбу");
        
        Console.WriteLine("\n3. Зоопарковые вольеры:");
        ZooEnclosure<Bird> birdEnclosure = new ZooEnclosure<Bird>();
        birdEnclosure.AddInhabitant(new Bird("Чижик", 1, 12.0, true));
        birdEnclosure.AddInhabitant(new Bird("Соловей", 2, 18.0, true));
        birdEnclosure.DisplayInhabitants();
        
        Console.WriteLine("\n4. Система учета животных:");
        AnimalRegistry<Dog> dogRegistry = new AnimalRegistry<Dog>();
        dogRegistry.RegisterAnimal(dog1);
        dogRegistry.RegisterAnimal(dog2);
        dogRegistry.DisplayRegistry();
        
        AnimalRegistry<Fish> fishRegistry = new AnimalRegistry<Fish>();
        fishRegistry.RegisterAnimal(fish1);
        fishRegistry.RegisterAnimal(fish2);
        fishRegistry.DisplayRegistry();
        
        Console.WriteLine("\n5. Поиск животных:");
        Dog foundDog = dogCollection.FindByName("Барсик");
        if (foundDog != null)
        {
            Console.WriteLine("Найдена собака:");
            foundDog.MakeSound();
            foundDog.Fetch("палку");
        }
        
        Console.WriteLine("\n6. Индексатор:");
        Console.WriteLine($"Первая собака в коллекции: {dogCollection[0].Name}");

        Console.WriteLine("\n7. Общая коллекция всех животных:");
        AnimalCollection<Animal> generalCollection = new AnimalCollection<Animal>();
        generalCollection.Add(dog1);
        generalCollection.Add(cat1);
        generalCollection.Add(bird1);
        generalCollection.Add(fish1);
        generalCollection.DisplayAll();
        generalCollection.MakeAllSounds();
        
        Console.WriteLine("\n8. Удаление из коллекции:");
        dogCollection.Remove(dog1);
        dogCollection.DisplayAll();
        Console.WriteLine($"Количество собак после удаления: {dogCollection.Count}");
````

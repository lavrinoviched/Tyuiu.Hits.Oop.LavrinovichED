<h1 style="color:DodgerBlue">Делегаты, лямбда-выражения и события<T></h1>



Делегаты, лямбда-выражения и события являются важными концепциями в C#, которые позволяют работать с методами как с объектами, писать краткий код и управлять реакциями на события. Рассмотрим их на примере гипотетических классов `Transport`, `Car`, и `Bike`.

### Делегаты

**Делегат** — это тип безопасного указателя на метод, похожий на функцию в C или C++. Делегаты в C# используются для передачи методов как параметров, а также для создания обратных вызовов.

**Пример:**

Создадим делегат, который может ссылаться на методы, выводящие информацию о транспортном средстве:

```csharp
public delegate void DisplayDelegate();

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Model: {Model}, Manufacturer: {Manufacturer}");
    }
}
```

Используем делегат:

```csharp

        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        // Присваиваем делегату метод DisplayInfo
        DisplayDelegate displayCar = car.DisplayInfo;
        DisplayDelegate displayBike = bike.DisplayInfo;

        // Вызов методов через делегаты
        displayCar();
        displayBike();

    }
```

### Лямбда-выражения

**Лямбда-выражения** позволяют вам писать анонимные методы в более краткой форме. Это часто используется в LINQ и асинхронном программировании.

**Пример:**

```csharp
using System;
using System.Collections.Generic;
using System.Linq;


 
List<Transport> transports = new List<Transport>
{
    new Car("Model S", "Tesla", 4),
    new Bike("Speedster", "Giant", true),
    new Car("Model 3", "Tesla", 4),
};

// Используем лямбда-выражение для фильтрации транспортных средств
var teslaCars = transports.OfType<Car>().Where(c => c.Manufacturer == "Tesla");

foreach (var car in teslaCars)
{
    car.DisplayInfo();
}


```

### События

**События** используются для оповещения подписчиков о случившихся действиях, предоставляя более высокоуровневый способ взаимодействия между различными частями кода.

**Пример:**

```csharp
public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }

    public delegate void TransportHandler(string message);
    public event TransportHandler TransportUpdated;

    public void UpdateModel(string newModel)
    {
        Model = newModel;
        // Вызываем событие после обновления модели
        TransportUpdated?.Invoke($"Model updated to {newModel}");
    }
}



Transport transport = new Transport { Model = "Old Model", Manufacturer = "Generic" };
// Подписываем метод на событие
transport.TransportUpdated += OnTransportUpdated;
// Обновляем модель; это вызовет событие
transport.UpdateModel("New Model");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```

### Объяснения и комментарии:

- **Делегаты**: Тип данных, который можно использовать для хранения ссылок на методы. Полезно для обратных вызовов и передачи методов.

- **Лямбда-выражения**: Более краткая запись анонимных методов, часто используется в LINQ для фильтрации, проекций и агрегаций.

- **События**: Механизм уведомления объектов об изменениях состояния или других действиях. Позволяют подписываться на события в одной части программы и реакцию на них в другой.

Эти механизмы делают программу гибкой и более структурированной, обеспечивая четкое отделение логики и упрощая повторное использование кода.

````
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void DisplayDelegate();
public delegate void TransportEventHandler(string message);

public class Transport
{
    public string Model { get; set; }
    public string Manufacturer { get; set; }
    public event TransportEventHandler TransportUpdated;

    public Transport(string model, string manufacturer)
    {
        Model = model;
        Manufacturer = manufacturer;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Модель: {Model}, Производитель: {Manufacturer}");
    }

    public void UpdateModel(string newModel)
    {
        string oldModel = Model;
        Model = newModel;
        TransportUpdated?.Invoke($"Модель изменена с '{oldModel}' на '{newModel}'");
    }

    protected virtual void OnTransportUpdated(string message)
    {
        TransportUpdated?.Invoke(message);
    }
}

public class Car : Transport
{
    public int NumberOfDoors { get; set; }

    public Car(string model, string manufacturer, int numberOfDoors)
        : base(model, manufacturer)
    {
        NumberOfDoors = numberOfDoors;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Количество дверей: {NumberOfDoors}");
    }

    public void UpdateDoors(int newNumberOfDoors)
    {
        int oldDoors = NumberOfDoors;
        NumberOfDoors = newNumberOfDoors;
        OnTransportUpdated($"Количество дверей изменено с {oldDoors} на {newNumberOfDoors}");
    }
}

public class Bike : Transport
{
    public bool HasCarrier { get; set; }

    public Bike(string model, string manufacturer, bool hasCarrier)
        : base(model, manufacturer)
    {
        HasCarrier = hasCarrier;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Имеет багажник: {HasCarrier}");
    }

    public void ToggleCarrier()
    {
        HasCarrier = !HasCarrier;
        OnTransportUpdated($"Багажник {(HasCarrier ? "добавлен" : "удален")}");
    }
}
        
        Car car = new Car("Model S", "Tesla", 4);
        Bike bike = new Bike("Speedster", "Giant", true);

        DisplayDelegate displayCar = car.DisplayInfo;
        DisplayDelegate displayBike = bike.DisplayInfo;

        Console.WriteLine("Вызов через делегаты:");
        displayCar();
        Console.WriteLine();
        displayBike();

        DisplayDelegate multiDelegate = car.DisplayInfo;
        multiDelegate += bike.DisplayInfo;
        
        Console.WriteLine("\nМногоадресный делегат:");
        multiDelegate();
        
        List<Transport> transports = new List<Transport>
        {
            new Car("Model S", "Tesla", 4),
            new Bike("Speedster", "Giant", true),
            new Car("Model 3", "Tesla", 4),
            new Bike("Urban", "Trek", false),
            new Car("Camry", "Toyota", 4)
        };

        var teslaCars = transports.Where(t => t.Manufacturer == "Tesla");
        
        Console.WriteLine("Транспорт Tesla:");
        foreach (var transport in teslaCars)
        {
            transport.DisplayInfo();
        }

        var sortedTransports = transports.OrderBy(t => t.Manufacturer).ThenBy(t => t.Model);
        
        Console.WriteLine("\nОтсортированный транспорт:");
        foreach (var transport in sortedTransports)
        {
            transport.DisplayInfo();
        }

        var transportInfo = transports.Select(t => new 
        { 
            Type = t.GetType().Name,
            t.Manufacturer, 
            t.Model 
        });
        
        Console.WriteLine("\nАнонимные типы:");
        foreach (var info in transportInfo)
        {
            Console.WriteLine($"Тип: {info.Type}, Производитель: {info.Manufacturer}, Модель: {info.Model}");
        }

        Console.WriteLine("\n=== СОБЫТИЯ ===");
        
        car.TransportUpdated += OnTransportUpdate;
        bike.TransportUpdated += OnTransportUpdate;

        car.TransportUpdated += (msg) => Console.WriteLine($"[ЛЯМБДА] {msg}");

        Console.WriteLine("Обновление машины:");
        car.UpdateModel("Model X");
        car.UpdateDoors(2);

        Console.WriteLine("\nОбновление велосипеда:");
        bike.UpdateModel("Mountain Pro");
        bike.ToggleCarrier();

        car.TransportUpdated -= OnTransportUpdate;

        Console.WriteLine("\nПосле отписки от событий машины:");
        car.UpdateModel("Model Y");
        
        var groupedTransports = transports.GroupBy(t => t.Manufacturer);
        
        foreach (var group in groupedTransports)
        {
            Console.WriteLine($"\nПроизводитель: {group.Key}");
            foreach (var transport in group)
            {
                Console.WriteLine($"  - {transport.Model}");
            }
        }

        var manufacturers = transports.Select(t => t.Manufacturer).Distinct();
        Console.WriteLine($"\nУникальные производители: {string.Join(", ", manufacturers)}");

        var firstBike = transports.FirstOrDefault(t => t is Bike);
        if (firstBike != null)
        {
            Console.WriteLine($"\nПервый велосипед: {firstBike.Model}");
        }
````

````
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void DisplayAnimalDelegate();
public delegate void AnimalEventHandler(string message);

public class Animal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public event AnimalEventHandler AnimalUpdated;

    public Animal(string name, int age)
    {
        Name = name;
        Age = age;
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age} лет");
    }

    public void UpdateName(string newName)
    {
        string oldName = Name;
        Name = newName;
        AnimalUpdated?.Invoke($"Имя изменено с '{oldName}' на '{newName}'");
    }

    protected virtual void OnAnimalUpdated(string message)
    {
        AnimalUpdated?.Invoke(message);
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    
    public Dog(string name, int age, string breed)
        : base(name, age)
    {
        Breed = breed;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed}");
    }

    public void UpdateBreed(string newBreed)
    {
        string oldBreed = Breed;
        Breed = newBreed;
        OnAnimalUpdated($"Порода изменена с '{oldBreed}' на '{newBreed}'");
    }
}

public class Cat : Animal
{
    public string Color { get; set; }
    
    public Cat(string name, int age, string color)
        : base(name, age)
    {
        Color = color;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Цвет: {Color}");
    }

    public void UpdateColor(string newColor)
    {
        string oldColor = Color;
        Color = newColor;
        OnAnimalUpdated($"Цвет изменен с '{oldColor}' на '{newColor}'");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }
    
    public Bird(string name, int age, bool canFly)
        : base(name, age)
    {
        CanFly = canFly;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Умеет летать: {CanFly}");
    }

    public void ToggleFlight()
    {
        CanFly = !CanFly;
        OnAnimalUpdated($"Способность летать изменена на: {CanFly}");
    }
}

public class Fish : Animal
{
    public string WaterType { get; set; }
    
    public Fish(string name, int age, string waterType)
        : base(name, age)
    {
        WaterType = waterType;
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Тип воды: {WaterType}");
    }

    public void UpdateWaterType(string newWaterType)
    {
        string oldWaterType = WaterType;
        WaterType = newWaterType;
        OnAnimalUpdated($"Тип воды изменен с '{oldWaterType}' на '{newWaterType}'");
    }
}

    static void OnAnimalUpdate(string message)
    {
        Console.WriteLine($"[СОБЫТИЕ] {message}");
    }
        
        Dog dog = new Dog("Бобик", 3, "Овчарка");
        Cat cat = new Cat("Мурка", 2, "Рыжий");
        Bird bird = new Bird("Кеша", 1, true);

        DisplayAnimalDelegate displayDog = dog.DisplayInfo;
        DisplayAnimalDelegate displayCat = cat.DisplayInfo;
        DisplayAnimalDelegate displayBird = bird.DisplayInfo;

        Console.WriteLine("Вызов через делегаты:");
        displayDog();
        Console.WriteLine();
        displayCat();
        Console.WriteLine();
        displayBird();

        DisplayAnimalDelegate multiDelegate = dog.DisplayInfo;
        multiDelegate += cat.DisplayInfo;
        multiDelegate += bird.DisplayInfo;
        
        Console.WriteLine("\nМногоадресный делегат:");
        multiDelegate();
        
        List<Animal> animals = new List<Animal>
        {
            new Dog("Бобик", 3, "Овчарка"),
            new Cat("Мурка", 2, "Рыжий"),
            new Bird("Кеша", 1, true),
            new Fish("Немо", 1, "Морская"),
            new Dog("Шарик", 4, "Дворняжка"),
            new Cat("Васька", 3, "Черный")
        };

        var dogs = animals.Where(a => a is Dog);
        
        Console.WriteLine("Собаки:");
        foreach (var animal in dogs)
        {
            animal.DisplayInfo();
        }

        var youngAnimals = animals.Where(a => a.Age < 3);
        
        Console.WriteLine("\nМолодые животные:");
        foreach (var animal in youngAnimals)
        {
            animal.DisplayInfo();
        }

        var sortedAnimals = animals.OrderBy(a => a.GetType().Name).ThenBy(a => a.Name);
        
        Console.WriteLine("\nОтсортированные животные:");
        foreach (var animal in sortedAnimals)
        {
            animal.DisplayInfo();
        }

        var animalInfo = animals.Select(a => new 
        { 
            Type = a.GetType().Name,
            a.Name, 
            a.Age 
        });
        
        Console.WriteLine("\nАнонимные типы:");
        foreach (var info in animalInfo)
        {
            Console.WriteLine($"Тип: {info.Type}, Имя: {info.Name}, Возраст: {info.Age}");
        }

        Console.WriteLine("\n=== СОБЫТИЯ ===");
        
        dog.AnimalUpdated += OnAnimalUpdate;
        cat.AnimalUpdated += OnAnimalUpdate;
        bird.AnimalUpdated += OnAnimalUpdate;

        dog.AnimalUpdated += (msg) => Console.WriteLine($"[ЛЯМБДА] {msg}");
        cat.AnimalUpdated += (msg) => Console.WriteLine($"[КОТ] {msg}");

        Console.WriteLine("Обновление собаки:");
        dog.UpdateName("Рекс");
        dog.UpdateBreed("Доберман");

        Console.WriteLine("\nОбновление кота:");
        cat.UpdateName("Барсик");
        cat.UpdateColor("Серый");

        Console.WriteLine("\nОбновление птицы:");
        bird.UpdateName("Гоша");
        bird.ToggleFlight();

        dog.AnimalUpdated -= OnAnimalUpdate;

        Console.WriteLine("\nПосле отписки от событий собаки:");
        dog.UpdateName("Дружок");
        
        var groupedAnimals = animals.GroupBy(a => a.GetType().Name);
        
        foreach (var group in groupedAnimals)
        {
            Console.WriteLine($"\nТип: {group.Key}");
            foreach (var animal in group)
            {
                Console.WriteLine($"  - {animal.Name}");
            }
        }

        var animalNames = animals.Select(a => a.Name).Distinct();
        Console.WriteLine($"\nУникальные имена: {string.Join(", ", animalNames)}");

        var oldestAnimal = animals.OrderByDescending(a => a.Age).FirstOrDefault();
        if (oldestAnimal != null)
        {
            Console.WriteLine($"\nСамое старое животное: {oldestAnimal.Name}, {oldestAnimal.Age} лет");
        }

        var averageAge = animals.Average(a => a.Age);
        Console.WriteLine($"Средний возраст животных: {averageAge:F1} лет");
````
